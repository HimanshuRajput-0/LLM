# How LLM Works:

LLM stands for Large Language Models:

GPT stands for Generative Pretrained Transformer.

This generative word creates a big difference between GPT and Search Engines.

Search engines just scrolls or finds the best outcome for our search.

But GPT's can generate the output according to our search.


GPT's just generates the next sequences using pretrained data and works on transformer architecture.

**Basically LLM's are just transformers who are Generative in nature and works on pretrained data.**

# Architecture of the Transformer.

**Step 1: Encoding.**


means Tokenization - split the text into seperate parts and maps them on particular number which is always fixed.

Tokenization varies for model to model.

Can see the examples of this on the keytokenizer website.

**We can also make our tokenizer by using transformers library in python by hugging face.**

We don't require any GPU for tokenizer.

```python
!pip install --upgrade transformers
```

```python
from transformers import AutoTokenizer
```

We are going to use gemma and it is a gated model so we need to provide it our hugging face access token as per google guidance for our security don't share this with anyone.

```python
import os

os.environ["HF_TOKEN"] = ""
```

```python
tokenizer = AutoTokenizer.from_pretrained('google/gemma-4-E4B-it')
```

```python
tokenizer("Himanshu")
```

```python
tokenizer("hii, my name is Himanshu Rajput")
```

Tokenization is possible because of vocab size.

```python
from openai import OpenAI
import os
os.environ["OPENAI_API_KEY"] = ""
```

```python
client = OpenAI()
response = client.embeddings.create(
    input = "My name is Himanshu",
    model = "text-embedding-3-small"
)

print(response.data[0].embedding)
```

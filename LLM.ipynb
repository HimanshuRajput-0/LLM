# How LLM Works:

LLM stands for Large Language Models:

GPT stands for Generative Pretrained Transformer.

This generative word creates a big difference between GPT and Search Engines.

Search engines just scrolls or finds the best outcome for our search.

But GPT's can generate the output according to our search.


GPT's just generates the next sequences using pretrained data and works on transformer architecture.

**Basically LLM's are just transformers who are Generative in nature and works on pretrained data.**
